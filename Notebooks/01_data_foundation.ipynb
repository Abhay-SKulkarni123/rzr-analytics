## 4. Data Quality Audit

Before analysing lead quality, the dataset needs to be checked for issues that could affect the results.

The audit focuses on missing values, duplicate records, identifier uniqueness, and the values present in key categorical fields. These checks help distinguish genuine characteristics of the data from potential data-quality problems.

### Missing Values

```python
missing = (df.isna().sum().to_frame("missing_count"))
missing["missing_pct"] = (missing["missing_count"] / len(df) * 100)
missing = missing.sort_values("missing_count", ascending=False)

missing
```

### Key Categorical Value Audit

```python
categorical_columns = ["CallStatus", "WidgetName", "PublisherZoneName", "PublisherCampaignName", "AdvertiserCampaignName"]

for column in categorical_columns:
    print(f"\n{column}")
    print("-" * len(column))
    print(df[column].value_counts(dropna=False))
```

## Final Conclusion

The initial data audit establishes the structure, completeness, and reliability of the dataset before performing lead quality analysis.

The dataset contains **3,021 lead records across 24 columns**, covering the period from **April 2, 2009 to September 30, 2009**. The data contains a mixture of lead attributes, acquisition information, campaign details, geographic information, and outcome-related fields.

### Data Quality Findings

`VendorLeadID` is the intended lead identifier. There are **8 missing identifier values**, while the remaining non-null identifiers are not completely unique, with **3,012 unique non-null IDs across 3,021 records**. This should be considered when using the identifier for record-level analysis.

No completely duplicated rows were identified in the dataset, reducing the likelihood that exact duplicate records will materially distort aggregate lead-level metrics.

Several fields contain substantial missingness. The most significant examples are:

- `IP Address`: **100% missing**
- `CallStatus`: **70.84% missing**
- `Keyword`: **67.59% missing**
- `AddressScore`: **61.24% missing**
- `Referring Keyword String`: **58.13% missing**
- `PhoneScore`: **53.89% missing**

The missing `CallStatus` values are particularly important because `CallStatus` contains the advertiser's lead outcome information and is used to identify `Closed` leads. These records should therefore not be automatically classified as either successful or unsuccessful outcomes without additional evidence.

The `IP Address` field is completely missing and therefore does not provide usable analytical information in its current form.

### Dataset Structure

The categorical audit shows that the dataset contains:

- **14 WidgetName variants**
- **2 PublisherZoneName values**
- **2 PublisherCampaignName values**
- **2 AdvertiserCampaignName values**

The limited number of publisher and advertiser campaign variants makes these variables suitable for direct segment comparisons in the subsequent analysis.

### Lead Outcome Structure

`CallStatus` contains both positive and negative outcome states as well as intermediate progression states. There are **245 Closed leads**, while a large proportion of records have no recorded `CallStatus`.

This means that the dataset provides a usable basis for analyzing observed Closed outcomes, but the missing outcome information must remain an explicit limitation throughout the analysis.

### Overall Conclusion

The dataset is sufficiently structured to proceed with lead quality analysis, but several data-quality limitations need to be carried forward. In particular, missing outcome information, incomplete lead identifiers, and missing scoring/acquisition fields can affect the interpretation of segment-level results.

The most important analytical principle established from this audit is that missing values should be preserved and interpreted according to their business meaning rather than being automatically converted into positive or negative outcomes.

With the data structure and limitations established, the next analysis focuses on the advertiser's primary success measure, **Closed Rate**, and examines how it changes over time and across relevant lead and acquisition segments.
